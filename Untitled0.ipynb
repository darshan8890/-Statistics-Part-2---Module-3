Q1.  What is hypothesis testing in statistics?
- Hypothesis testing is a statistical method used to make decisions or inferences about a population based on sample data. It helps determine whether there is enough evidence to reject a certain assumption (hypothesis) about a population parameter.

Q2. What is the null hypothesis, and how does it differ from the alternative hypothesis?

- Null Hypothesis = The null hypothesis represents the default or initial assumption that there is no effect, no difference, or no relationship in the population. It assumes that any observed changes or differences in the sample are due to chance.

- Alternative Hypothesis = The alternative hypothesis is the statement that contradicts the null hypothesis. It suggests that there is a significant effect, difference, or relationship in the population.

Q3. What is the significance level in hypothesis testing, and why is it important?

- The significance level (α) is the probability of rejecting the null hypothesis (H0) when it is actually true. It represents the risk of making a Type I error (false positive) in hypothesis testing.
- It sets the threshold for deciding whether the observed results are statistically significant.
- A lower α (e.g., 0.01) makes the test more stringent, reducing false positives but increasing the risk of missing a true effect (Type II error).

Q4. What does a P-value represent in hypothesis testing?

- The p-value is the probability of obtaining a test statistic at least as extreme as the observed data, assuming that the null hypothesis (H0) is true. It helps determine the strength of evidence against H0.

Q5. How do you interpret the P-value in hypothesis testing?

- The p-value tells us how likely it is to observe the given sample results (or something more extreme) if the null hypothesis (H0) is true.

Q6. What are Type 1 and Type 2 errors in hypothesis testing ?

- Type 1 Error = Occurs when we reject the null hypothesis (H0) even though it is actually true. It’s like a false alarm—detecting an effect when none exists. The probability of making a Type I error is represented by the significance level (a).

- Type 2 Error = Occurs when we fail to reject the null hypothesis (H0) even though it is actually false. It’s like missing a real effect—failing to detect something that actually exists. The probability of making a Type II error is represented by β.

Q7. What is the difference between a one-tailed and a two-tailed test in hypothesis testing?

- One-Tailed Hypothesis Testing = A one-tailed test is used when we are only interested in one direction of change (either an increase or a decrease). It is More statistically powerful when the effect is in the expected direction.

- Two-Tailed Hypothesis Testing = A two-tailed test is used when we are interested in both directions (increase or decrease). It is Used when any difference (positive or negative) is meaningful.

Q8. What is the Z-test, and when is it used in hypothesis testing?
- A Z-test is a statistical test used to determine whether the mean of a sample is significantly different from a known population mean, assuming the population follows a normal distribution. It is based on the Z-score, which measures how many standard deviations a data point is from the mean.
- Types of Z-Tests = One-Sample Z-Test, Two-Sample Z-Test

Q9. How do you calculate the Z-score, and what does it represent in hypothesis testing?

- A Z-score (or standard score) tells us how many standard deviations a data point (or sample mean) is from the population mean.

Q10. What is the T-distribution, and when should it be used instead of the normal distribution ?

- The T-distribution (Student’s T-distribution) is a probability distribution used for estimating population parameters when the sample size is small and/or the population standard deviation (σ) is unknown.
- It is similar to the normal distribution but has fatter tails, meaning it accounts for more variability in small samples.
- As the sample size increases, the T-distribution approaches the normal distribution.

Q13. What is the relationship between Z-test and T-test in hypothesis testing?

- Both tests are used to determine whether a sample mean significantly differs from a population mean or to compare two sample means. They help decide whether to reject the null hypothesis (H0) in favor of the alternative hypothesis (Ha).

Q14. What is a confidence interval, and how is it used to interpret statistical results?
- A confidence interval (CI) is a range of values used to estimate an unknown population parameter (e.g., mean or proportion) based on sample data. It provides a measure of uncertainty around the estimate and is often expressed at a confidence level, such as 95% or 99%.
- A confidence interval is calculated as: Confidence Interval = Estimate +- Margin of Error.

Q15.  What is the margin of error, and how does it affect the confidence interval?
- The margin of error (MOE) is the range within which the true population parameter is expected to lie, based on sample data. It quantifies the uncertainty in an estimate and is influenced by the sample size, confidence level, and variability in the data.
- The margin of error is calculated as: MOE = Critical Value * Standard Error.
- The margin of error determines the precision of a confidence interval. A smaller MOE results in a more precise estimate, while a larger MOE increases uncertainty. It depends on sample size, confidence level, and data variability.

Q16. How is Bayes' Theorem used in statistics, and what is its significance?
- Bayes' Theorem is a fundamental concept in probability theory and statistics that describes how to update the probability of a hypothesis based on new evidence or data.
- It provides a way to calculate the posterior probability of an event, given prior knowledge (prior probability) and the likelihood of the evidence.

Q17. What is the Chi-square distribution, and when is it used?
- The Chi-square distribution is a continuous probability distribution that is widely used in statistical inference. It is the distribution of the sum of the squares of k independent standard normal random variables, where k is a positive integer.
- This means that if you have a set of k independent normal variables with mean 0 and variance 1, the sum of their squares will follow a Chi-square distribution with k degrees of freedom (df).

Q18. What is the Chi-square goodness of fit test, and how is it applied ?
- The Chi-square goodness of fit test is a statistical test used to determine whether a set of observed frequencies matches a set of expected frequencies, based on a specific hypothesis.
- It is used when you have categorical data and want to test if the distribution of your data fits a particular theoretical distribution (e.g., uniform distribution, normal distribution, etc.).
- The test compares the observed frequencies in each category with the expected frequencies if the hypothesis were true. If there is a significant difference between the observed and expected frequencies, the null hypothesis is rejected.

Q19. What is the F-distribution, and when is it used in hypothesis testing?
- The F-distribution is a continuous probability distribution that arises from the ratio of two independent Chi-square distributions, each divided by their respective degrees of freedom. It is commonly used in statistical inference, especially in variance analysis and regression analysis.
- The F-distribution is primarily used in hypothesis testing related to variances and comparison of multiple groups. Some common applications include:  Analysis of Variance (ANOVA), Testing the Equality of Variances, Regression Analysis (F-test for Overall Significance).

Q20. What is an ANOVA test, and what are its assumptions ?
- ANOVA (Analysis of Variance) is a statistical test used to compare the means of three or more independent groups to determine if there is a significant difference between them. It is commonly used when you want to see if different groups (e.g., different treatments, categories, or populations) have different average outcomes on a dependent variable.
- The fundamental idea behind ANOVA is to analyze the variance within groups and between groups to see if the variation between the groups is larger than the variation within each group. If the between-group variance is much larger than the within-group variance, it suggests that at least one of the group means is different.
- Types of ANOVA: One-Way ANOVA, Two-Way ANOVA.


Q21. What are the different types of ANOVA tests?
- There are several types of ANOVA tests used in statistical analysis, each designed for different scenarios and experimental designs. The most common types includes:
- One-Way ANOVA (Single-Factor ANOVA)
- Two-Way ANOVA (Two-Factor ANOVA)
- Three-Way ANOVA (Three-Factor ANOVA)
- Repeated Measures ANOVA
- Multivariate Analysis of Variance (MANOVA)
- Mixed-Design ANOVA

Q22. What is the F-test, and how does it relate to hypothesis testing?
- The F-test is a statistical test used to compare the variances of two or more populations or groups to assess if they are significantly different. It is based on the F-distribution, which arises when comparing the ratio of two sample variances, each divided by their respective degrees of freedom.
-  The F-test can be used in various hypothesis testing contexts, but its main applications are in:
- Testing the equality of variances between two populations or samples.
- Analysis of Variance (ANOVA), where it is used to compare the means of three or more groups by testing if the variability between groups is significantly larger than the variability within groups.

```python
# Q1. Write a Python program to perform a Z-test for comparing a sample mean to a known population mean and interpret the results.

import numpy as np
from scipy import stats

sample_data = np.array([50, 55, 52, 58, 60, 53, 56, 59, 62, 54])
population_mean = 55
population_std = 5
sample_size = len(sample_data)

sample_mean = np.mean(sample_data)

z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

print(f"Sample Mean: {sample_mean}")
print(f"Z-score: {z_score}")
print(f"P-value: {p_value}")

alpha = 0.05

if p_value < alpha:
    print("\nConclusion: Reject the null hypothesis (the sample mean is significantly different from the population mean).")
else:
    print("\nConclusion: Fail to reject the null hypothesis (the sample mean is not significantly different from the population mean).")
```

```python
# Q5. Create a Python function that calculates and visualizes Type 1 and Type 2 errors during hypothesis testing.

def calculate_errors(population_mean, sample_data, alpha=0.05):
    sample_mean = np.mean(sample_data)
    sample_std = np.std(sample_data, ddof=1)
    n = len(sample_data)

    z_stat = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
    critical_value = stats.norm.ppf(1 - alpha/2)

    type_1_error_region = (-np.inf, -critical_value), (critical_value, np.inf)
    type_2_error_region = (population_mean - critical_value, population_mean + critical_value)

    z_values = np.linspace(-5, 5, 1000)
    y_values = stats.norm.pdf(z_values, 0, 1)

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.plot(z_values, y_values, label="Standard Normal Distribution")

    ax.fill_between(z_values, y_values, where=(z_values <= -critical_value) | (z_values >= critical_value), color='red', alpha=0.5, label="Rejection Region (Type 1 Error)")
    ax.fill_between(z_values, y_values, where=(z_values >= -critical_value) & (z_values <= critical_value), color='yellow', alpha=0.5, label="Non-Rejection Region (Type 2 Error)")

    ax.axvline(x=z_stat, color='black', linestyle='--', label=f'Z-stat = {z_stat:.2f}')
    ax.axvline(x=-critical_value, color='blue', linestyle='--', label="Critical Value (-1.96)")
    ax.axvline(x=critical_value, color='blue', linestyle='--', label="Critical Value (1.96)")

    plt.title("Type 1 and Type 2 Errors during Hypothesis Testing")
    plt.xlabel("Z-value")
    plt.ylabel("Probability Density")
    plt.legend()
    plt.show()

    return z_stat

np.random.seed(0)
sample_data = np.random.normal(loc=55, scale=10, size=100)
population_mean = 50
calculate_errors(population_mean, sample_data)
```

```python
# Q9. Write a Python function to calculate the confidence interval for a sample mean and explain its significance.

import numpy as np
import scipy.stats as stats

def confidence_interval(sample_data, confidence_level=0.95):
    sample_mean = np.mean(sample_data)
    sample_size = len(sample_data)
    sample_std = np.std(sample_data, ddof=1)

    standard_error = sample_std / np.sqrt(sample_size)

    critical_value = stats.t.ppf((1 + confidence_level) / 2, df=sample_size - 1)

    margin_of_error = critical_value * standard_error

    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    return (lower_bound, upper_bound)

sample_data = np.random.normal(loc=50, scale=10, size=30)
confidence_level = 0.95
ci = confidence_interval(sample_data, confidence_level)
print(f"Confidence Interval: {ci}")
```

```python
# Q10. Write a Python program to calculate the margin of error for a given confidence level using sample data.

import numpy as np
import scipy.stats as stats

def margin_of_error(sample_data, confidence_level=0.95):
    sample_mean = np.mean(sample_data)
    sample_size = len(sample_data)
    sample_std = np.std(sample_data, ddof=1)

    standard_error = sample_std / np.sqrt(sample_size)

    critical_value = stats.t.ppf((1 + confidence_level) / 2, df=sample_size - 1)

    margin_of_error = critical_value * standard_error

    return margin_of_error

sample_data = np.random.normal(loc=50, scale=10, size=30)
confidence_level = 0.95
moe = margin_of_error(sample_data, confidence_level)
print(f"Margin of Error: {moe}")
```

```python
# Q11. Implement a Bayesian inference method using Bayes' Theorem in Python and explain the process.

def bayesian_inference(prior, likelihood, evidence):
    posterior = (likelihood * prior) / evidence
    return posterior

prior = 0.01
likelihood = 0.95
false_positive_rate = 0.05

evidence = (likelihood * prior) + (false_positive_rate * (1 - prior))

posterior = bayesian_inference(prior, likelihood, evidence)

print(f"Posterior Probability P(H|E): {posterior:.4f}")
```

```python
# Q16. Implement an F-test using Python to compare the variances of two random samples.

import numpy as np
import scipy.stats as stats

sample1 = np.random.normal(loc=0, scale=1, size=50)
sample2 = np.random.normal(loc=0, scale=2, size=50)

var1 = np.var(sample1, ddof=1)
var2 = np.var(sample2, ddof=1)

f_stat = var1 / var2
dof1 = len(sample1) - 1
dof2 = len(sample2) - 1

p_value = 1 - stats.f.cdf(f_stat, dof1, dof2)

print(f"F-statistic: {f_stat}, P-value: {p_value}")
```

```python
# Q17. Write a Python program to perform an ANOVA test to compare means between multiple groups and interpret the results.

import numpy as np
import scipy.stats as stats

group1 = np.random.normal(loc=50, scale=10, size=30)
group2 = np.random.normal(loc=55, scale=10, size=30)
group3 = np.random.normal(loc=60, scale=10, size=30)

f_stat, p_value = stats.f_oneway(group1, group2, group3)

print(f"F-statistic: {f_stat}, P-value: {p_value}")

if p_value < 0.05:
    print("There is a significant difference between the means of the groups.")
else:
    print("There is no significant difference between the means of the groups.")
```

```python
# Q19. Write a Python function to check the assumptions (normality, independence, and equal variance) for ANOVA.

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

def check_anova_assumptions(groups):
    for i, group in enumerate(groups, start=1):
        stat, p_value = stats.shapiro(group)
        print(f"Normality Test for Group {i}: W-statistic={stat}, p-value={p_value}")
        if p_value < 0.05:
            print(f"Group {i} does not follow normal distribution.")
        else:
            print(f"Group {i} follows normal distribution.")

    print("\nAssumption of Independence is assumed based on experimental design.\n")

    stat, p_value = stats.levene(*groups)
    print(f"Equal Variance Test (Levene's Test): Statistic={stat}, p-value={p_value}")
    if p_value < 0.05:
        print("The groups have unequal variances.")
    else:
        print("The groups have equal variances.")

    plt.figure(figsize=(8, 6))
    plt.boxplot(groups)
    plt.title('Boxplot of Groups')
    plt.ylabel('Value')
    plt.show()

group1 = np.random.normal(loc=10, scale=5, size=30)
group2 = np.random.normal(loc=12, scale=5, size=30)
group3 = np.random.normal(loc=14, scale=5, size=30)

check_anova_assumptions([group1, group2, group3])
```

```python
# Q20. D Perform a two-way ANOVA test using Python to study the interaction between two factors and visualize the results.


import numpy as np
import pandas as pd
import scipy.stats as stats
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
factor1 = np.random.choice(['A', 'B'], size=100)
factor2 = np.random.choice(['X', 'Y'], size=100)
value = np.random.normal(loc=10, scale=5, size=100)

data = pd.DataFrame({'Factor1': factor1, 'Factor2': factor2, 'Value': value})

model = stats.f_oneway(
    data[data['Factor1'] == 'A'][data['Factor2'] == 'X']['Value'],
    data[data['Factor1'] == 'A'][data['Factor2'] == 'Y']['Value'],
    data[data['Factor1'] == 'B'][data['Factor2'] == 'X']['Value'],
    data[data['Factor1'] == 'B'][data['Factor2'] == 'Y']['Value']
)

print(f"F-statistic: {model.statistic}, P-value: {model.pvalue}")

plt.figure(figsize=(8, 6))
sns.boxplot(x='Factor1', y='Value', hue='Factor2', data=data)
plt.title('Interaction between Factor1 and Factor2')
plt.show()
```

```python
# Q21. Write a Python program to visualize the F-distribution and discuss its use in hypothesis testing.

import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

dfn = 5
dfd = 2

x = np.linspace(0, 5, 1000)
f_pdf = stats.f.pdf(x, dfn, dfd)

plt.figure(figsize=(8, 6))
plt.plot(x, f_pdf, label=f'F-distribution df1={dfn}, df2={dfd}')
plt.fill_between(x, f_pdf, alpha=0.4)
plt.title('F-distribution')
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Q22. Perform a one-way ANOVA test in Python and visualize the results with boxplots to compare group means.

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

np.random.seed(0)
group_1 = np.random.normal(loc=10, scale=5, size=30)
group_2 = np.random.normal(loc=15, scale=5, size=30)
group_3 = np.random.normal(loc=20, scale=5, size=30)

f_statistic, p_value = stats.f_oneway(group_1, group_2, group_3)

print("F-statistic:", f_statistic)
print("p-value:", p_value)

data = [group_1, group_2, group_3]
group_labels = ['Group 1', 'Group 2', 'Group 3']

plt.figure(figsize=(8, 6))
sns.boxplot(data=data, width=0.5, palette='Set2')
plt.xticks(ticks=np.arange(len(group_labels)), labels=group_labels)
plt.title('Boxplots of Groups')
plt.ylabel('Values')
plt.show()

if p_value < 0.05:
    print("There is a significant difference between the groups.")
else:
    print("There is no significant difference between the groups.")
```

```python
# Q23. Simulate random data from a normal distribution, then perform hypothesis testing to evaluate the means.

import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

np.random.seed(0)
sample_size = 100
sample_data = np.random.normal(loc=5, scale=2, size=sample_size)

hypothesized_mean = 5
t_statistic, p_value = stats.ttest_1samp(sample_data, hypothesized_mean)

print(f"T-statistic: {t_statistic}")
print(f"P-value: {p_value}")

plt.hist(sample_data, bins=20, edgecolor='black', alpha=0.7)
plt.axvline(hypothesized_mean, color='r', linestyle='--', label='Hypothesized Mean')
plt.title('Simulated Data Distribution')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.legend()
plt.show()

if p_value < 0.05:
    print("Reject the null hypothesis: The sample mean is significantly different from 5.")
else:
    print("Fail to reject the null hypothesis: The sample mean is not significantly different from 5.")
```

```python
# Q24. Perform a hypothesis test for population variance using a Chi-square distribution and interpret the results.

import numpy as np
from scipy import stats

np.random.seed(0)
sample_size = 30
sample_data = np.random.normal(loc=5, scale=2, size=sample_size)

hypothesized_variance = 4
sample_variance = np.var(sample_data, ddof=1)

chi_square_statistic = (sample_size - 1) * sample_variance / hypothesized_variance
degrees_of_freedom = sample_size - 1

p_value_lower = stats.chi2.cdf(chi_square_statistic, degrees_of_freedom)
p_value_upper = 1 - p_value_lower
p_value = 2 * min(p_value_lower, p_value_upper)

print(f"Chi-square statistic: {chi_square_statistic}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("Reject the null hypothesis: The population variance is significantly different from the hypothesized value.")
else:
    print("Fail to reject the null hypothesis: The population variance is not significantly different from the hypothesized value.")
```

```python
# Q25. Write a Python script to perform a Z-test for comparing proportions between two datasets or groups.

import numpy as np
from scipy import stats

x1 = 50
n1 = 200
x2 = 60
n2 = 250

p1 = x1 / n1
p2 = x2 / n2

p = (x1 + x2) / (n1 + n2)

z = (p1 - p2) / np.sqrt(p * (1 - p) * (1 / n1 + 1 / n2))

p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print(f"Z-statistic: {z}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("Reject the null hypothesis: The proportions are significantly different.")
else:
    print("Fail to reject the null hypothesis: The proportions are not significantly different.")
```

```python
# Q26. Implement an F-test for comparing the variances of two datasets, then interpret and visualize the results.

import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(0)
sample_size = 100
data1 = np.random.normal(loc=5, scale=2, size=sample_size)
data2 = np.random.normal(loc=5, scale=3, size=sample_size)

var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)

f_statistic = var1 / var2
df1 = sample_size - 1
df2 = sample_size - 1

p_value = 1 - stats.f.cdf(f_statistic, df1, df2)

print(f"F-statistic: {f_statistic}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("Reject the null hypothesis: The variances are significantly different.")
else:
    print("Fail to reject the null hypothesis: The variances are not significantly different.")

plt.figure(figsize=(8, 6))
sns.boxplot(data=[data1, data2], width=0.5, palette='Set2')
plt.xticks([0, 1], ['Dataset 1', 'Dataset 2'])
plt.title('Boxplot Comparison of Two Datasets')
plt.ylabel('Values')
plt.show()
```

```python
# Q27. Perform a Chi-square test for goodness of fit with simulated data and analyze the results.

import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

observed = np.array([12, 15, 10, 14, 16, 13])

expected = np.array([15, 15, 15, 15, 15, 15])
expected = expected * (observed.sum() / expected.sum())

chi_square_statistic, p_value = stats.chisquare(observed, expected)

print(f"Chi-square statistic: {chi_square_statistic}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("Reject the null hypothesis: The observed data does not follow the expected distribution.")
else:
    print("Fail to reject the null hypothesis: The observed data follows the expected distribution.")

labels = ['1', '2', '3', '4', '5', '6']
x = np.arange(len(labels))

plt.bar(x - 0.2, observed, width=0.4, label='Observed', align='center')
plt.bar(x + 0.2, expected, width=0.4, label='Expected', align='center')
plt.xticks(x, labels)
plt.xlabel('Dice Faces')
plt.ylabel('Frequency')
plt.title('Observed vs Expected Frequencies (Dice Roll)')
plt.legend()
plt.show()
```
